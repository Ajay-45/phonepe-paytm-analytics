```python
import json
import pandas as pd

file_path = "/content/npcI_upi_apps_historical_2022_2026.json"

with open(file_path, "r", encoding="utf-8") as f:
    data = json.load(f)

df = pd.DataFrame(data["results"])

print("Records:", len(df))
print("Columns:", df.columns.tolist())

df.head()
```

```python
print("Total records:", len(df))

print("\nYears:")
print(sorted(df["year"].astype(str).unique()))

print("\nMonths:")
print(df.groupby("year")["month"].nunique())

print("\nPhonePe records:")
print(df[df["application_name"].str.contains("Phone Pe", case=False, na=False)].shape[0])

print("\nPaytm records:")
print(df[df["application_name"].str.contains("Paytm", case=False, na=False)].shape[0])
```

```python
print("PhonePe names:")
print(
    df[df["application_name"].str.contains("phone", case=False, na=False)]
    ["application_name"]
    .value_counts()
)

print("\nPaytm names:")
print(
    df[df["application_name"].str.contains("paytm", case=False, na=False)]
    ["application_name"]
    .value_counts()
)
```

```python


def standardize_app_name(name):
    name = str(name).strip().lower()


    if "phonepe" in name or "phone pe" in name:
        return "PhonePe"


    if "paytm" in name:
        return "Paytm"

    return str(name).strip()


df["application_name"] = df["application_name"].apply(standardize_app_name)

print(df["application_name"].value_counts().head(20))
```

```python
with open(file_path, "r", encoding="utf-8") as f:
    original_data = json.load(f)

df = pd.DataFrame(original_data["results"])


# Standardize only PhonePe and Paytm names
def standardize_app_name(name):
    original = str(name).strip()
    check = original.lower()

    if "phonepe" in check or "phone pe" in check:
        return "PhonePe"

    if "paytm" in check:
        return "Paytm"

    return original


df["application_name"] = df["application_name"].apply(standardize_app_name)


# Save the cleaned historical dataset
output_path = "/content/upi_apps_historical_cleaned_2022_2026.csv"
df.to_csv(output_path, index=False)

print("Saved:", output_path)
print("Records:", len(df))
print("PhonePe:", (df["application_name"] == "PhonePe").sum())
print("Paytm:", (df["application_name"] == "Paytm").sum())
```
